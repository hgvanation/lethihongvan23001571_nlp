```python
text = "Dự kiến, số thu ngân sách từ thuế thu nhập cá nhân còn khoảng 84.477 tỷ đồng, giảm 21.000 tỷ mỗi năm so với hiện hành. Số người còn phải nộp thuế là 2,21 triệu người, giảm 2,18 triệu người (49,66%) do chuyển từ bậc 1 sang diện không chịu thuế."
text
```

```python
text_temp = "(49,66%)" # ==> [(, 49, ",", 66, %, ")"]
def complete_split(text):
  # print(1111)
  def split_punc(text_input):
    for p in string.punctuation:
      if p in text_input:
        t =  text_input.split(p)
        return [t[0], p, t[1]]
    return [text_input]
  test_out1 = split_punc(text)
  # print(test_out1)
  while True:
    result = []
    flag = False
    for i in test_out1:
      if len(i) > 1:
        temp = split_punc(i)
        result+= temp
        if len(temp) > 1 : flag = True
      else:
        result.append(i)
    if not flag:
      break
    test_out1 = result
  if "" in test_out1: test_out1.remove("")
  return test_out1


complete_split(text_temp)
```

```python
def process(text):
  split_text = text.split()
  output = []
  for word in split_text:
    temp = []
    temp = complete_split(word)
    if len(temp) == 0:
      output.append(word)
    else:
      output += temp
  return output

process_text = process(text)
process_text
```
