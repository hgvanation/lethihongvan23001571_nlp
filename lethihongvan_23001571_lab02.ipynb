# Count Vectorization

```python
corpus = [
    "I love NLP.",
"I love programming. programming programming programming programming",
"NLP is a subfield of AI.",
]
print(corpus)
```

```python
corpus = [i.lower() for i in corpus]
```

```python
tokenized_corpus = [
    tokenizer.tokenize(doc) for doc in corpus
]
tokenized_corpus
```

```python
vocab = set()
for t in tokenized_corpus:
  vocab = vocab | set(t)
vocab
```

```python
vocab = sorted(list(vocab))
vocab
```

```python
def build_vector(tokenized_text, vocab):
  count = {k:0 for k in vocab}
  for token in tokenized_text:
    count[token] += 1
  vector = [0 for _ in range(len(vocab))]
  for i in range(len(vocab)):
    vector[i] = count[vocab[i]]
  return vector
for tokenized_text in tokenized_corpus:
  print(tokenized_text)
  print(build_vector(tokenized_text, vocab))
  print(vocab)
  print("============")
```

```python
import re
from collections import Counter
import pandas as pd

def basic_tokenize(text):
    text = text.lower()
    for p in ['.', ',', '?', '!']:
        text = text.replace(p, f' {p} ')
    tokens = text.split()
    return tokens

def advanced_tokenize(text):
    text = text.lower()
    pattern = r"\w+|[^\w\s]"
    tokens = re.findall(pattern, text)
    return tokens

sample_sentences = [
    "Hello, world! This is a test.",
    "NLP is fascinating... isn't it?",
    "Let's see how it handles 123 numbers and punctuation!"
]
for i, sent in enumerate(sample_sentences, 1):
    print(f"\nCâu {i}: \"{sent}\"")
    print("Basic Tokenize   :", basic_tokenize(sent))
    print("Advanced Tokenize:", advanced_tokenize(sent))

ud_ewt_sample = (
    "Al - Zaman : American forces landed in Amiriya today to take back control of the town. "
    "The troops were met with fierce resistance from local insurgents. "
    "What if Google expanded on its search engine? It has been reported that the company plans "
    "to launch new artificial intelligence tools to transform healthcare."
)

print("\n--- KẾT QUẢ SO SÁNH TRÊN UD_English-EWT ---")
print("Tổng số token (Basic Tokenize)   :", len(basic_tokenize(ud_ewt_sample)))
print("Tổng số token (Advanced Tokenize):", len(advanced_tokenize(ud_ewt_sample)))

class CountVectorizerFromScratch:
    def __init__(self, tokenizer_func=advanced_tokenize):
        self.tokenizer = tokenizer_func
        self.vocabulary = {}

    def fit(self, corpus):
        unique_tokens = set()
        for doc in corpus:
            tokens = self.tokenizer(doc)
            unique_tokens.update(tokens)

        sorted_tokens = sorted(list(unique_tokens))
        self.vocabulary = {token: idx for idx, token in enumerate(sorted_tokens)}
        return self.vocabulary

    def transform(self, corpus):
        matrix = []
        vocab_size = len(self.vocabulary)

        for doc in corpus:
            vector = [0] * vocab_size
            tokens = self.tokenizer(doc)
            token_counts = Counter(tokens)

            for token, count in token_counts.items():
                if token in self.vocabulary:
                    idx = self.vocabulary[token]
                    vector[idx] = count

            matrix.append(vector)

        return matrix

print("\n--- KẾT QUẢ MINH HỌA PHẦN 2 ---")
demo_corpus = [
    "I love NLP.",
    "I love programming.",
    "NLP is a subfield of AI."
]

demo_vectorizer = CountVectorizerFromScratch(tokenizer_func=advanced_tokenize)
demo_vocab = demo_vectorizer.fit(demo_corpus)
demo_matrix = demo_vectorizer.transform(demo_corpus)

print("Từ điển (Vocabulary):")
print(demo_vocab)

print("\nMa trận Document-Term:")
df_demo = pd.DataFrame(demo_matrix, columns=sorted(demo_vocab.keys()), index=[f"Văn bản {i+1}" for i in range(len(demo_corpus))])
print(df_demo)

file_path = "fox_news_text.txt"
with open(file_path, "r", encoding="utf-8") as f:
    raw_content = f.read()

corpus_fox_news = [para.strip() for para in raw_content.split("\n\n") if para.strip()]

fox_vectorizer = CountVectorizerFromScratch(tokenizer_func=advanced_tokenize)
fox_vocab = fox_vectorizer.fit(corpus_fox_news)
fox_dtm = fox_vectorizer.transform(corpus_fox_news)

df_fox = pd.DataFrame(
    fox_dtm,
    columns=sorted(fox_vocab.keys()),
    index=[f"Đoạn {i+1}" for i in range(len(corpus_fox_news))]
)

print("\n--- THỐNG KÊ TRÊN DỮ LIỆU FOX NEWS ---")
print(f"Tổng số đoạn văn (tài liệu): {len(corpus_fox_news)}")
print(f"Kích thước từ điển (Vocabulary Size): {len(fox_vocab)}")
print(f"Kích thước Ma trận (Rows x Cols): {df_fox.shape}")

print("\nHiển thị 5 đoạn văn đầu tiên và 10 từ đầu tiên trong Từ điển:")
print(df_fox.iloc[:5, :10])
```
